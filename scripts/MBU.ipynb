```python
import pandas as pd
```

```python
df_mbu = pd.read_excel('/content/MBU_Entry_Status_District_Wise 15-07-2026 12-10-17 PM.xlsx')
```

```python
df_mbu.columns = df_mbu.iloc[2]
```

```python
df_mbu = df_mbu.iloc[3:]
df_mbu.reset_index(drop=True , inplace=True)
```

```python
df_mbu.head()
```

```python
df_mbu = df_mbu.iloc[1:]
df_mbu.reset_index(drop=True , inplace=True)
```

```python
df_mbu.head()
```

```python
columns_to_aggregate = [
    'District Name',
    'MBU Pending (Age 5-15)',
    'MBU Pending (Age 15 and above)',
    'Total Students AADHAAR Provided'
]

df_selected_agg = df_mbu[columns_to_aggregate].copy()

# Convert specified columns to numeric
for col in columns_to_aggregate[1:]:
    df_selected_agg[col] = pd.to_numeric(df_selected_agg[col], errors='coerce')

# Group by 'District Name' and sum the selected numeric columns
df_final_aggregated = df_selected_agg.groupby('District Name', as_index=False).sum()

# Display the aggregated DataFrame
display(df_final_aggregated.head())
```

```python
df_final_aggregated['total pending'] = df_final_aggregated['MBU Pending (Age 5-15)'] + df_final_aggregated['MBU Pending (Age 15 and above)']
display(df_final_aggregated.head())
```

```python
import numpy as np

# Calculate 'Total Pending %', handling division by zero
df_final_aggregated['Total Pending %'] = np.where(
    df_final_aggregated['Total Students AADHAAR Provided'] != 0,
    (df_final_aggregated['total pending'] / df_final_aggregated['Total Students AADHAAR Provided']) * 100,
    0  # Assign 0% if 'Total Students AADHAAR Provided' is zero
).round(2)
display(df_final_aggregated.head())
```
